```python
import requests
import json
import folium
from datetime import datetime
import os

url = "https://map.stockholmarchipelagotrail.com/data/geojson/pois.geojson"

# Hämta GeoJSON
response = requests.get(url)
data = response.json()

# Filtrera fram hamnar
harbours = [
    feature for feature in data["features"]
    if feature.get("properties", {}).get("category") == "harbour"
]

# Skapa karta
m = folium.Map(location=[59.5, 18.8], zoom_start=8)

# Lägg till hamnar med klickbara SAT-länkar
for h in harbours:
    lon, lat = h["geometry"]["coordinates"]
    props = h["properties"]
    poi_id = props.get("id")
    name = props.get("name")

    link = f"http://map.stockholmarchipelagotrail.com/?id={poi_id}"
    linkjson = f"http://map.stockholmarchipelagotrail.com/?id={poi_id}"

def sat_link_json(poi_id):
    return f"https://map.stockholmarchipelagotrail.com/api/objects/{poi_id}"
    
    popup_html = f"""
    <b>{name}</b><br>
    ID: {poi_id}<br>
    <a href="{link}" target="_blank">Öppna i SAT-kartan</a>
    """

    folium.Marker(
        location=[lat, lon],
        popup=folium.Popup(popup_html, max_width=300),
        tooltip=name
    ).add_to(m)

# Skapa output-mapp om den saknas
os.makedirs("output", exist_ok=True)

# Datumformat YYYYMMDD
date_str = datetime.now().strftime("%Y%m%d")

# Filnamn
filename = f"output/481_hamnar_SAT_{date_str}.html"

# Spara kartan
m.save(filename)

print(f"Kartan exporterad till: {filename}")
```
