```python
import time
import datetime  
start_time = time.time()
start_str = datetime.datetime.now().strftime("%Y-%m-%d %H:%M")
print(f"Started: {start_str}")
```

```python
import requests

url_poi_concordance = "https://map.stockholmarchipelagotrail.com/data/geojson/poi-concordance.json"
data = requests.get(url_poi_concordance).json()

wikidata_ids = [
    k.replace("wikidata:", "")
    for k in data["satIdOf"].keys()
    if k.startswith("wikidata:")
]

print(f"{len(wikidata_ids)} Wikidata-objekt")
print(wikidata_ids[:10])
```

```python
osm_ids = [
    k.replace("osm:", "")
    for k in data["satIdOf"].keys()
    if k.startswith("osm:")
]
print(f"{len(osm_ids)} OSM-objekt")
print(osm_ids[:10])
```
